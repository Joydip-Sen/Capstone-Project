```python
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
```

```python
df=pd.read_csv('flats.csv')
df.sample(5)
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.drop(columns=['link','property_id'],inplace=True)
```

```python
df.head()
```

```python
df.rename(columns={'area':'price_per_sqft'},inplace=True)
df.head()
```

```python
df['price_per_sqft'].isnull().sum()
```

```python
df=df[~df['price_per_sqft'].isnull()]
df.shape
```

```python
df.head()
```

```python
df['price_per_sqft']=df['price_per_sqft'].str.split('/').str.get(0).str.replace('₹','').str.replace(',','').str.strip().astype('float')
```

```python
df.head()
```

```python
df['price'].str.split(' ').str.get(1).value_counts()
```

```python
def convert_price(price):
  if type(price) == float:
    return price;
  else:
    if price[1]=='Lac':
      return round(float(price[0])/100,2);
    else:
      return round(float(price[0]),2)
```

```python
df['price']=df['price'].str.split(' ').apply(convert_price)
```

```python
df.head()
```

```python
df['society'].value_counts().shape
```

```python
df['society']=df['society'].str.lower()
```

```python
df['society'].value_counts()
```

```python
import re
df['society']=df['society'].apply(lambda name:re.sub(r'\d+(\.\d+)?\s?★','',str(name).strip()))
df['society']
```
